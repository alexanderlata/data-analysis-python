Реализуйте функцию `stacked_hist(df, column, group, title, xlabel, ax=None)`, которая рисует гистограмму целых значений со слоями по группам. Возвращаемое значение: объект Axes со столбцом шириной 1 с центром в каждом целом значении column, разбитым на слои по группам: высота столбца — число ДТП с этим значением, слой — число ДТП группы; заголовок title, подпись оси X xlabel, подпись оси Y «Число ДТП», легенда с названиями групп. Проверка в ноутбуке: ДТП области по дням недели и освещению через stacked_hist; печатаются день недели самого высокого столбца и его высота, оба вводятся в тест.

```python
def stacked_hist(df, column, group, title, xlabel, ax=None):
    """
    Рисует гистограмму целых значений со слоями по группам.

    Гистограмма семинара 8 отвечала на вопрос «сколько», гистограмма с
    накоплением отвечает «сколько и из чего»: вечерний пик ДТП состоит из
    тёмного времени. `hue` и `multiple='stack'` делают это одной строкой; без
    `discrete=True` столбцы не совпадают с целыми значениями, без
    `multiple='stack'` слои накладываются друг на друга, и высоты лгут.

    Аргументы:
        df: Таблица.
        column: Имя столбца с целыми значениями, строка: час, день недели, число
                участников.
        group: Имя столбца с группами, строка: освещение, тяжесть.
        title: Заголовок, строка.
        xlabel: Подпись оси X, строка.
        ax: Оси для рисования; если не переданы, функция создаёт свои: fig, ax =
            plt.subplots(figsize=(8, 4)).

    Возвращаемое значение:
        Объект Axes со столбцом шириной 1 с центром в каждом целом значении
        column, разбитым на слои по группам: высота столбца — число ДТП с этим
        значением, слой — число ДТП группы; заголовок title, подпись оси X
        xlabel, подпись оси Y «Число ДТП», легенда с названиями групп. Таблица
        не меняется.
    """

    pass
```

```python
def stacked_hist_tests():
    df = pd.DataFrame({'v': [1, 1, 2, 2, 2, 4], 'g': ['a', 'b', 'a', 'a', 'b', 'a']})
    before = df.copy()
    ax = stacked_hist(df, 'v', 'g', 'Заголовок', 'Значение')
    assert isinstance(ax, plt.Axes), 'функция должна возвращать Axes'
    tops = {}
    for p in ax.patches:
        if p.get_height() > 0:
            center = round(p.get_x() + p.get_width() / 2, 3)
            tops[center] = max(tops.get(center, 0), round(p.get_y() + p.get_height(), 3))
    assert tops == {1: 2, 2: 3, 4: 1}, 'столбец на каждое целое значение, вершина — число ДТП со значением, слои друг на друге'
    assert all(abs(p.get_width() - 1) < 0.01 for p in ax.patches), 'ширина столбца — единица: discrete=True'
    assert sorted(t.get_text() for t in ax.get_legend().get_texts()) == ['a', 'b'], 'легенда с названиями групп'
    assert ax.get_title() == 'Заголовок' and ax.get_xlabel() == 'Значение' and ax.get_ylabel() == 'Число ДТП'
    assert df.equals(before), 'таблица изменилась'
    fig, own = plt.subplots()
    assert stacked_hist(df, 'v', 'g', 'Заголовок', 'Значение', ax=own) is own, 'переданные оси должны использоваться'
    plt.close('all')

    print('Все тесты прошли успешно!')
```

```python
def density_by_tests():
    df = pd.DataFrame({'v': [1, 2, 2, 3, 3, 3, 4, 5, 8, 10], 'g': ['a'] * 8 + ['b'] * 2})
    before = df.copy()
    ax = density_by(df, 'v', 'g', 'Заголовок', 'Значение')
    assert isinstance(ax, plt.Axes), 'функция должна возвращать Axes'
    assert len(ax.lines) == 2, 'на осях должны быть две линии, по одной на группу, без заливки'
    areas = [float(np.trapezoid(line.get_ydata(), line.get_xdata())) for line in ax.lines]
    assert all(abs(area - 1) < 0.05 for area in areas), 'площадь под каждой кривой должна быть 1: common_norm=False'
    assert sorted(t.get_text() for t in ax.get_legend().get_texts()) == ['a', 'b'], 'легенда с названиями групп'
    assert ax.get_title() == 'Заголовок' and ax.get_xlabel() == 'Значение' and ax.get_ylabel() == 'Плотность'
    assert df.equals(before), 'таблица изменилась'
    fig, own = plt.subplots()
    assert density_by(df, 'v', 'g', 'Заголовок', 'Значение', ax=own) is own, 'переданные оси должны использоваться'
    plt.close('all')

    print('Все тесты прошли успешно!')
```

```python
def corr_heat(df, columns, title, ax=None):
    """
    Рисует тепловую карту корреляций числовых столбцов.

    Корреляционная матрица — одна команда и один график для всех числовых
    столбцов отчёта. `corr()` по всей таблице падает на тексте, список столбцов
    нужен и ему, и читателю: координаты и номера в матрице ничего не значат.
    Последовательная палитра прячет знак корреляции, расходящаяся с нулём в
    середине показывает его цветом.

    Аргументы:
        df: Таблица.
        columns: Список имён числовых столбцов.
        title: Заголовок, строка.
        ax: Оси для рисования; если не переданы, функция создаёт свои: fig, ax =
            plt.subplots(figsize=(6, 5)).

    Возвращаемое значение:
        Объект Axes с тепловой картой матрицы df[columns].corr(): расходящаяся
        палитра с нейтральным цветом в нуле, красное — положительная связь,
        синее — отрицательная; в каждой клетке значение с двумя знаками после
        запятой; подписи делений осей — имена столбцов; заголовок title. Таблица
        не меняется.
    """

    pass
```

```python
def corr_heat_tests():
    df = pd.DataFrame({'a': [1, 2, 3, 4], 'b': [1, 3, 2, 4], 'c': [2, 4, 3, 1], 'd': [2, 2, 3, 1], 'name': list('wxyz')})
    before = df.copy()
    ax = corr_heat(df, ['a', 'b', 'c'], 'Заголовок')
    assert isinstance(ax, plt.Axes), 'функция должна возвращать Axes'
    expected = df[['a', 'b', 'c']].corr().values.ravel().round(2).tolist()
    cells = np.asarray(ax.collections[0].get_array()).ravel().round(2).tolist()
    assert cells == expected, 'клетки — корреляции столбцов из списка, лишние столбцы не рисуются'
    assert [t.get_text() for t in ax.texts] == [f'{v:.2f}' for v in expected], 'в каждой клетке значение с двумя знаками: annot=True, fmt=".2f"'
    assert [t.get_text() for t in ax.get_xticklabels()] == ['a', 'b', 'c'], 'подписи делений — имена столбцов'
    zero = ax.collections[0].to_rgba(0)
    assert min(zero[:3]) > 0.8, 'цвет нуля нейтральный: расходящаяся палитра с center=0'
    assert ax.get_title() == 'Заголовок'
    assert df.equals(before), 'таблица изменилась'
    fig, own = plt.subplots()
    assert corr_heat(df, ['a', 'b', 'c'], 'Заголовок', ax=own) is own, 'переданные оси должны использоваться'
    plt.close('all')

    print('Все тесты прошли успешно!')
```

```python
def report_page_tests():
    df = pd.DataFrame({'p': [1, 1, 2, 2, 2, 3, 3, 3], 'f': [True, False, True, True, False, True, True, True],
                       'g': ['a', 'b', 'a', 'a', 'b', 'a', 'b', 'b'], 'u': [1, 2, 3, 4, 5, 6, 7, 8], 'w': [2, 1, 4, 3, 6, 5, 8, 7]})
    before = df.copy()
    fig = report_page(df, 'p', 'f', 'g', ['u', 'w'], 'Общий заголовок')
    assert isinstance(fig, plt.Figure), 'функция должна возвращать Figure'
    panels = [ax for ax in fig.axes if ax.get_label() != '<colorbar>']
    assert len(panels) == 4, 'четыре панели, не считая шкалы цвета'
    def panel(left, top):
        found = [ax for ax in panels if (ax.get_position().x0 < 0.5) == left and (ax.get_position().y0 > 0.4) == top]
        assert len(found) == 1, 'панели в сетке 2 × 2'
        return found[0]
    top_left, top_right, bottom_left, bottom_right = panel(True, True), panel(False, True), panel(True, False), panel(False, False)
    tops = {}
    for p in top_left.patches:
        if p.get_height() > 0:
            center = round(p.get_x() + p.get_width() / 2)
            tops[center] = max(tops.get(center, 0), round(p.get_y() + p.get_height(), 3))
    assert tops == {1: 2, 2: 3, 3: 3} and top_left.get_legend() is not None, 'слева сверху гистограмма с накоплением по группам'
    assert np.allclose(top_right.lines[0].get_ydata(), [0.5, 2 / 3, 1]) and len(top_right.collections) >= 1, 'справа сверху доля по периодам с полосой'
    bars = {t.get_text(): round(float(p.get_height()), 3) for t, p in zip(bottom_left.get_xticklabels(), bottom_left.patches)}
    assert bars == {'a': 1.0, 'b': 0.5} and len(bottom_left.lines) + len(bottom_left.collections) >= 1, 'слева снизу доля по группам с усами'
    assert len(bottom_right.collections) >= 1 and len(bottom_right.texts) == 4 and '1.00' in [t.get_text() for t in bottom_right.texts], 'справа снизу корреляции с подписями'
    titles = [top_left.get_title(), top_right.get_title(), bottom_left.get_title(), bottom_right.get_title()]
    assert titles == ['Число ДТП по периодам', 'Доля по периодам', 'Доля по группам', 'Корреляции'], 'заголовки панелей как в условии'
    assert fig._suptitle.get_text() == 'Общий заголовок', 'общий заголовок через suptitle'
    assert df.equals(before), 'таблица изменилась'
    plt.close('all')

    print('Все тесты прошли успешно!')
```

Реализуйте функцию `missing_bars(df, title, ax=None)`, которая рисует доли пропусков по столбцам таблицы. Аргументы: `df` — любая таблица; `title` — заголовок, строка; `ax` — оси для рисования; если не переданы, функция создаёт свои: fig, ax = plt.subplots(figsize=(8, 4)). Возвращаемое значение: объект Axes с горизонтальными столбцами по столбцам таблицы, в которых есть пропуски: длина столбца — доля пропусков, самый дырявый столбец сверху, столбцы без пропусков не показываются; заголовок title, подпись оси X «Доля пропусков». Таблица не меняется.

Подсказка: `shares = df.isna().mean()`, затем `shares[shares > 0].sort_values().plot.barh(ax=ax)`: `barh` рисует первый элемент снизу, поэтому сортировка по возрастанию ставит наибольшую долю наверх; путь через убывание и `ax.invert_yaxis()` тесты тоже принимают.

Проверка в ноутбуке: пропуски в сырой таблице участников области через missing_bars; печатаются столбец с наибольшей долей пропусков и его доля, столбец с наименьшей долей среди показанных и его доля, вводятся в тест.

```python
def missing_bars(df, title, ax=None):
    """
    Рисует доли пропусков по столбцам таблицы.

    Раздел «Пропуски» отчёта этапа 2 — первый график после загрузки данных. В
    семинаре 6 та же сводка была таблицей, здесь она становится графиком;
    функция `missing_report` из того семинара в этом ноутбуке недоступна, доли
    считаются заново одной строкой. Коды 95, 96, 99 в стаже — скрытые пропуски:
    на графике их нет, пока их не заменить.

    Аргументы:
        df: Любая таблица.
        title: Заголовок, строка.
        ax: Оси для рисования; если не переданы, функция создаёт свои: fig, ax =
            plt.subplots(figsize=(8, 4)).

    Возвращаемое значение:
        Объект Axes с горизонтальными столбцами по столбцам таблицы, в которых
        есть пропуски: длина столбца — доля пропусков, самый дырявый столбец
        сверху, столбцы без пропусков не показываются; заголовок title, подпись
        оси X «Доля пропусков». Таблица не меняется.
    """

    pass
```

```python
def missing_bars_tests():
    df = pd.DataFrame({'a': [1, None, None, None], 'b': [None, 2, 3, 4], 'c': [1, 2, 3, 4], 'd': ['x', None, None, 'z']})
    before = df.copy()
    ax = missing_bars(df, 'Заголовок')
    assert isinstance(ax, plt.Axes), 'функция должна возвращать Axes'
    assert len(ax.patches) == 3, 'столбец на каждый столбец таблицы с пропусками; без пропусков не показываются'
    labels = [t.get_text() for t in ax.get_yticklabels()]
    pairs = list(zip(labels, [round(float(p.get_width()), 3) for p in ax.patches]))
    visual = pairs if ax.yaxis_inverted() else pairs[::-1]
    assert visual == [('a', 0.75), ('d', 0.5), ('b', 0.25)], 'самый дырявый столбец сверху, длина — доля пропусков'
    assert ax.get_title() == 'Заголовок' and ax.get_xlabel() == 'Доля пропусков'
    assert df.equals(before), 'таблица изменилась'
    fig, own = plt.subplots()
    assert missing_bars(df, 'Заголовок', ax=own) is own, 'переданные оси должны использоваться'
    plt.close('all')

    print('Все тесты прошли успешно!')
```

Возвращаемое значение: объект plotly.graph_objects.Figure: карта функцией points_map по строкам с dead_count > 0 внутри рамки, цвет точки по столбцу dead_count, в подсказке district, datetime и category, заголовок «ДТП с погибшими»; та же фигура записана в файл path через write_html с include_plotlyjs='cdn'. Подсказка: `fatal = df[df['dead_count'] > 0]`, затем `points_map(fatal, box, 'dead_count', ['district', 'datetime', 'category'], 'ДТП с погибшими')`, `fig.write_html(path, include_plotlyjs='cdn')`, `return fig`; раздел 5 ноутбука занятия о сохранении.

Проверка в ноутбуке: карта ДТП с погибшими Московской области в файл img/fatal-map.html через fatal_map; печатаются число точек с двумя и более погибшими и доля точек с одним погибшим, оба вводятся в тест.

```python
def fatal_map(df, box, path):
    """
    Рисует карту ДТП с погибшими и сохраняет её в HTML.

    Карта, которую можно отдать читателю без Python: файл открывается в
    браузере, подсказки и масштаб работают. Со встроенной библиотекой файл весит
    почти пять мегабайт даже для трёх точек; с `include_plotlyjs='cdn'` —
    килобайты, но нужен интернет, а подложке карты он нужен и так. Цвет по числу
    погибших выделяет редкие ДТП с несколькими жертвами.

    Аргументы:
        df: Таблица ДТП со столбцами lat, lon, dead_count, district, datetime,
            category.
        box: Кортеж из четырёх чисел (lat_min, lat_max, lon_min, lon_max): рамка
             области.
        path: Путь к файлу HTML, строка; папка уже существует: папку img/
              создаёт ячейка подключения ноутбука.

    Возвращаемое значение:
        Объект plotly.graph_objects.Figure: карта функцией points_map по строкам
        с dead_count > 0 внутри рамки, цвет точки по столбцу dead_count, в
        подсказке district, datetime и category, заголовок «ДТП с погибшими»; та
        же фигура записана в файл path через write_html с
        include_plotlyjs='cdn'. Таблица не меняется.
    """

    pass
```

```python
def fatal_map_tests():
    import os
    import tempfile
    df = pd.DataFrame({'lat': [51.0, 51.5, 50.5, 31.0, 0.0, 51.2], 'lon': [31.0, 31.5, 30.5, 51.0, 0.0, 31.8], 'dead_count': [1, 0, 2, 1, 3, 3],
                       'district': list('abcdef'), 'datetime': pd.to_datetime(['2024-01-0%d' % i for i in range(1, 7)]), 'category': ['k'] * 6})
    before = df.copy()
    path = os.path.join(tempfile.mkdtemp(), 'fatal.html')
    fig = fatal_map(df, (50, 52, 30, 32), path)
    assert len(fig.data) == 1 and fig.data[0].type in ('scattermap', 'scattermapbox'), 'одна группа точек на карте: цвет по числу погибших'
    trace = fig.data[0]
    assert len(trace.lat) == 3, 'только ДТП с погибшими внутри рамки'
    assert sorted(int(v) for v in trace.marker.color) == [1, 2, 3], 'цвет точки — число погибших'
    assert np.asarray(trace.customdata).shape[1] == 3, 'в подсказке район, дата и вид ДТП'
    assert fig.layout.title.text == 'ДТП с погибшими'
    assert os.path.exists(path), 'файл по пути path должен появиться'
    assert os.path.getsize(path) < 1_000_000, 'файл меньше мегабайта: include_plotlyjs="cdn"'
    html = open(path, encoding='utf-8').read()
    assert '<html' in html and 'cdn.plot.ly' in html, 'файл HTML с библиотекой plotly из интернета'
    assert df.equals(before), 'таблица изменилась'
    os.remove(path)
    os.rmdir(os.path.dirname(path))

    print('Все тесты прошли успешно!')
```
