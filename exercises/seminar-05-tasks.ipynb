### Б8. Тяжесть по группам в долях

Чтобы сравнивать группы между собой, нужны доли внутри группы: `crosstab` с `normalize='index'` считает строки по двум столбцам и сразу делит каждую строку на её сумму. Первый вопрос к такой таблице: как связана тяжесть ДТП с числом транспортных средств, столбец `vehicles_count`. ДТП с одним транспортным средством — это наезды на пешеходов, съезды с дороги, наезды на препятствие, а не столкновения.

Аргументы: `df` — таблица ДТП со столбцом severity; `column` — имя столбца для строк таблицы, строка. Возвращаемое значение: таблица, где строки — значения столбца column, столбцы — значения severity, в ячейках доля ДТП каждой тяжести внутри строки, округлённая до тысячных; сумма по строке равна единице.

Подсказка: раздел 6 ноутбука занятия: `pd.crosstab(df[column], df['severity'], normalize='index')` делит каждую строку на её сумму; `normalize=True` делил бы на всю таблицу; `round(3)` округляет всю таблицу.

Проверка в ноутбуке: тяжесть по числу транспортных средств для области, первые пять строк; в тест вводятся доли ДТП с погибшими при одном и при двух транспортных средствах.

```python
def severity_shares(df, column):
    """
    Строит таблицу долей тяжести ДТП внутри групп.

    В Б7 ячейки хранят суммы, и большие группы всегда выглядят страшнее
    маленьких. Чтобы сравнивать группы между собой, нужны доли внутри группы:
    `crosstab` с `normalize='index'` считает строки по двум столбцам и сразу
    делит каждую строку на её сумму. Первый вопрос к такой таблице: как связана
    тяжесть ДТП с числом транспортных средств, столбец `vehicles_count`. ДТП с
    одним транспортным средством — это наезды на пешеходов, съезды с дороги,
    наезды на препятствие, а не столкновения.

    Аргументы:
        df: Таблица ДТП со столбцом severity.
        column: Имя столбца для строк таблицы, строка.

    Возвращаемое значение:
        Таблица, где строки — значения столбца column, столбцы — значения
        severity, в ячейках доля ДТП каждой тяжести внутри строки, округлённая
        до тысячных; сумма по строке равна единице.
    """

    pass
```

```python
def severity_shares_tests():
    df = pd.DataFrame({'vehicles_count': [1, 1, 1, 1, 2, 2, 2, 2, 2, 2],
                       'severity': ['Тяжёлый', 'С погибшими', 'Легкий', 'Тяжёлый',
                                    'Тяжёлый', 'Тяжёлый', 'Тяжёлый', 'С погибшими', 'Легкий', 'Легкий']})
    result = severity_shares(df, 'vehicles_count')
    assert list(result.columns) == ['Легкий', 'С погибшими', 'Тяжёлый']
    assert result.loc[1].tolist() == [0.25, 0.25, 0.5]
    assert result.loc[2].tolist() == [0.333, 0.167, 0.5]
    assert round(float(result.loc[2].sum()), 2) == 1.0, 'доли в строке в сумме дают единицу'
    assert result.index.tolist() == [1, 2]

    print('Все тесты прошли успешно!')
```

```python
def fatal_share_by(df, column, min_count):
    """
    Считает долю ДТП с погибшими по группам, отбрасывая малые группы.

    Упражнение А6 считало долю по видам ДТП, и уже там пришлось отбирать
    массовые виды глазами: отбрасывание предмета с долей 0.25 стояло на 12 ДТП.
    По районам без порога первым оказывается город Бронницы: 0.41 при 39 ДТП.
    Порог по числу ДТП в группе делает рейтинг честным, и его надо применить к
    результату группировки, а не к исходной таблице: сначала посчитать в каждой
    группе число ДТП и долю, потом оставить группы с достаточным числом ДТП.

    Аргументы:
        df: Таблица ДТП со столбцом dead_count.
        column: Имя столбца, по которому группировать, строка.
        min_count: Наименьшее число ДТП в группе, целое число; порог включается.

    Возвращаемое значение:
        Series, где индекс — значения столбца column, значения — доля ДТП с хотя
        бы одним погибшим в группе, округлённая до тысячных; только группы, в
        которых не меньше min_count ДТП; по убыванию доли.
    """

    pass
```

```python
def fatal_share_by_tests():
    df = pd.DataFrame({'district': ['Клин', 'Клин', 'Клин', 'Клин', 'Химки', 'Химки', 'Химки', 'Дубна', 'Дубна'],
                       'dead_count': [1, 0, 0, 0, 1, 1, 0, 1, 1]})
    result = fatal_share_by(df, 'district', 3)
    assert isinstance(result, pd.Series), 'результат должен быть Series'
    assert result.index.tolist() == ['Химки', 'Клин'], 'Дубна с двумя ДТП отсекается порогом, Химки на пороге остаются'
    assert result.tolist() == [0.667, 0.25]
    assert fatal_share_by(df, 'district', 1).index.tolist() == ['Дубна', 'Химки', 'Клин']
    assert fatal_share_by(df, 'district', 4).tolist() == [0.25]
    assert len(fatal_share_by(df, 'district', 5)) == 0

    print('Все тесты прошли успешно!')
```
