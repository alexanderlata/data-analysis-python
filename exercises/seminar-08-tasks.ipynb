```python
def share_bars(rows, columns, order, title, ax=None):
    """
    Рисует столбцы с накоплением долей.

    Столбцы с накоплением отвечают на вопрос «из чего состоит»: тяжесть ДТП при
    разной погоде, виды ДТП по годам, исход по ролям участников. Доли считаются
    внутри группы, `normalize='index'`; порядок слоёв задаёт список, а не
    алфавит.

    Аргументы:
        rows: Series с группами, например погода или год; по строкам будущей
              сводки.
        columns: Series той же длины с тем, что раскладывается на доли, например
                 тяжесть; по столбцам сводки.
        order: Список значений columns в порядке слоёв снизу вверх; остальные
               значения не показываются.
        title: Заголовок, строка.
        ax: Оси для рисования; если не переданы, функция создаёт свои: fig, ax =
            plt.subplots(figsize=(8, 4)).

    Возвращаемое значение:
        Объект Axes, где на каждую группу один столбец, слои — доли значений
        внутри группы, сумма слоёв 1; заголовок title, подпись оси Y «Доля ДТП»,
        легенда со значениями из order, подписи групп без поворота. Входные
        Series не меняются.
    """

    pass
```

```python
def share_bars_tests():
    rows = pd.Series(['a', 'a', 'b', 'b'])
    columns = pd.Series(['x', 'y', 'y', 'y'])
    before = rows.copy(), columns.copy()
    ax = share_bars(rows, columns, ['x', 'y'], 'Заголовок')
    assert isinstance(ax, plt.Axes), 'функция должна возвращать Axes'
    assert len(ax.containers) == 2, 'слоёв столько, сколько значений в order'
    heights = [[round(p.get_height(), 3) for p in container] for container in ax.containers]
    assert heights == [[0.5, 0.0], [0.5, 1.0]], 'доли внутри группы: у a 0.5 и 0.5, у b 0 и 1'
    assert [round(p.get_y(), 3) for p in ax.containers[1]] == [0.5, 0.0], 'слои стоят друг на друге: stacked=True'
    assert [t.get_text() for t in ax.get_xticklabels()] == ['a', 'b']
    assert [t.get_text() for t in ax.get_legend().get_texts()] == ['x', 'y'], 'легенда в порядке order'
    assert ax.get_title() == 'Заголовок' and ax.get_ylabel() == 'Доля ДТП'
    assert len(share_bars(rows, columns, ['y'], 'Заголовок').containers) == 1, 'значения вне order не рисуются'
    assert rows.equals(before[0]) and columns.equals(before[1]), 'входные Series изменились'
    fig, own = plt.subplots()
    assert share_bars(rows, columns, ['x', 'y'], 'Заголовок', ax=own) is own, 'переданные оси должны использоваться'
    plt.close('all')

    print('Все тесты прошли успешно!')
```
