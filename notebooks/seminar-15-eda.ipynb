## Отчёт: кто его читает и как он устроен

Отчёт читает не автор и не программист, а юрист, руководитель, проверяющий — человек, которому нужен ответ, а не код. Читает он сверху вниз и часто только начало и конец: что за данные, можно ли им верить, что в них видно, что проверено и что из этого следует. Поэтому отчёт пишут так, чтобы его можно было прочитать, не читая кода: код — доказательство, текст — сам отчёт.

Структура постоянна: разделы этапа 2, `project/02-eda.md`, с его нумерацией, затем гипотезы этапа 3, `project/03-hypotheses.md`, затем выводы.

| Раздел | Что в нём | Вес в рубрике этапа |
|---|---|---|
| 1. Загрузка и первичный осмотр | данные и вопрос, размер, память, повторы, типы | 10 % этапа 2 |
| 2. Анализ пропущенных значений | доли пропусков, скрытые коды, связь пропусков с группами, решение | 15 % этапа 2 |
| 3. Описательные статистики | числовые и категориальные столбцы, время | 20 % этапа 2 |

Раздел гипотез помечен буквой Г: он стоит между разделами 6 и 7, а нумерация 1–7 взята из этапа 2, и её сохраняют. На защите этот же отчёт оценивается по рубрике `project/04-defense.md`: данные и качество, графики, гипотезы, выводы и ответы на вопросы.

Правило отчёта — **график — вывод — оговорка**. Под каждым графиком и тестом — вывод одним-двумя предложениями по памятке:

1. **Вывод** — ответ на вопрос с числом: «в тёмное время без освещения гибелью кончается 28,6 % ДТП, днём — 10,1 %», а не «на графике пять столбцов».
2. **Оговорка** — чего из этого не следует: выбросы и асимметрия, пропуски, смена учёта, связь — не причина, размер эффекта.

В начале каждого раздела — две строки: что в нём проверяет рубрика и частая ошибка. Частые ошибки взяты из ловушек семинаров 8–13: те же ошибки встречаются и в отчётах.

Функции не меняют входные данные. Функции графиков возвращают оси `Axes` или фигуру, и подпись, которую функция ставит по умолчанию, поправляют после вызова: `ax.set_ylabel(...)`. В ячейке — подключение библиотек и 32 функции; она ничего не печатает.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import plotly.express as px
import seaborn as sns
from scipy import stats

sns.set_theme(style='whitegrid')                                         # оформление графиков по умолчанию


def count_line(periods, title, xlabel, ax=None):   # число ДТП по периодам
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 4))
    counts = periods.value_counts().sort_index()
    counts.plot(ax=ax, marker='o')
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel('Число ДТП')
    return ax


def top_bars(values, n, title, ax=None):   # самые частые значения
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 4))
    counts = values.value_counts().head(n)[::-1]
    counts.plot.barh(ax=ax)
    ax.set_title(title)
    ax.set_xlabel('Число ДТП')
    return ax


def share_bars(rows, columns, order, title, ax=None):   # состав групп долями
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 4))
    shares = pd.crosstab(rows, columns, normalize='index')[order]
    shares.plot(kind='bar', stacked=True, rot=0, ax=ax)
    ax.set_title(title)
    ax.set_ylabel('Доля ДТП')
    return ax


def count_hist(values, title, xlabel, ax=None):   # гистограмма целых значений
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 4))
    sns.histplot(x=values, discrete=True, ax=ax)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel('Число ДТП')
    return ax


def box_by(df, value, by, order, title, ax=None):   # ящики по группам
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 4))
    sns.boxplot(data=df, x=by, y=value, order=order, ax=ax)
    ax.set_title(title)
    return ax


def density_by(df, column, group, title, xlabel, ax=None):   # плотности по группам
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 4))
    sns.kdeplot(data=df, x=column, hue=group, common_norm=False, ax=ax)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel('Плотность')
    return ax


def corr_heat(df, columns, title, ax=None):   # тепловая карта корреляций
    if ax is None:
        fig, ax = plt.subplots(figsize=(6, 5))
    corr = df[columns].corr()
    sns.heatmap(corr, cmap='coolwarm', center=0, annot=True, fmt='.2f', ax=ax)
    ax.set_title(title)
    return ax


def share_line(df, period, flag, title, xlabel, ax=None):   # доля по периодам
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 4))
    sns.lineplot(data=df, x=period, y=flag, ax=ax)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel('Доля')
    return ax


def mean_bars(df, group, value, title, xlabel, ylabel, ax=None):   # среднее по группам
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 4))
    sns.barplot(data=df, x=group, y=value, ax=ax)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    return ax


def hist_panels(df, column, by, title):   # гистограммы на панелях
    g = sns.displot(data=df, x=column, col=by, col_wrap=4, discrete=True, height=2.5)
    g.figure.suptitle(title)
    return g.figure


def points_map(df, box, color, hover, title):   # точки на карте
    lat_min, lat_max, lon_min, lon_max = box
    inside = df[df['lat'].between(lat_min, lat_max) & df['lon'].between(lon_min, lon_max)]
    fig = px.scatter_map(inside, lat='lat', lon='lon', color=color, hover_data=hover, zoom=7, height=600, title=title)
    return fig


def missing_bars(df, title, ax=None):   # доли пропусков по столбцам
    if ax is None:
        fig, ax = plt.subplots(figsize=(8, 4))
    shares = df.isna().mean()
    shares = shares[shares > 0].sort_values()
    shares.plot.barh(ax=ax)
    ax.set_title(title)
    ax.set_xlabel('Доля пропусков')
    return ax


def iqr_bounds(values, factor=1.5):   # границы IQR
    q1, q3 = values.quantile([0.25, 0.75])
    iqr = q3 - q1
    return q1 - factor * iqr, q3 + factor * iqr


def flag_outliers(values, method='iqr'):   # маска выбросов
    if method == 'iqr':
        lower, upper = iqr_bounds(values)
        return (values < lower) | (values > upper)
    if method == 'z':
        z = (values - values.mean()) / values.std()
        return z.abs() > 3
    if method == 'percentile':
        low, high = values.quantile([0.01, 0.99])
        return (values < low) | (values > high)
    raise ValueError(f'неизвестное правило: {method}')


def outlier_table(df, columns):   # выбросы по трём правилам
    table = pd.DataFrame(index=columns)
    for method in ['iqr', 'z', 'percentile']:
        table[method] = [flag_outliers(df[column], method).sum() for column in columns]
    return table


def missing_by_group(df, column, group):   # доля пропусков по группам
    shares = df[column].isna().groupby(df[group]).mean()
    return shares.sort_values(ascending=False)


def logic_checks(df):   # логические проверки ДТП
    victims = df['dead_count'] + df['injured_count']
    return pd.Series({
        'пострадавших больше участников': (victims > df['participants_count']).sum(),
        'машин больше участников': (df['vehicles_count'] > df['participants_count']).sum(),
        'машин нет': (df['vehicles_count'] == 0).sum(),
        'участников нет': (df['participants_count'] == 0).sum(),
    })


def duplicate_report(df, group):   # повторы по группам
    repeated = df.duplicated(keep=False)
    report = pd.DataFrame({
        'строк': df.groupby(group).size(),
        'в повторах': repeated.groupby(df[group]).sum(),
    })
    return report.sort_values('в повторах', ascending=False)


def normality_table(df, columns, alpha=0.05):   # тест Д’Агостино по столбцам
    table = pd.DataFrame(index=columns)
    table['n'] = [df[column].count() for column in columns]
    table['асимметрия'] = [df[column].skew() for column in columns]
    table['p'] = [stats.normaltest(df[column].dropna()).pvalue for column in columns]
    table['решение'] = np.where(table['p'] < alpha, 'отвергаем', 'не отвергаем')
    return table


def normality_report(df, columns, alpha=0.05):   # нормальность и выбор теста
    report = normality_table(df, columns, alpha)
    small = np.where(report['решение'] == 'отвергаем', 'непараметрический тест',
                     'мало данных: решать по Q–Q графику и асимметрии')
    report['что дальше'] = np.where(report['n'] >= 30, 't-тест', small)
    return report


def qq_plot(values, title, ax=None):   # Q–Q график
    if ax is None:
        fig, ax = plt.subplots(figsize=(5, 5))
    stats.probplot(values.dropna(), plot=ax)
    ax.set_title(title)
    ax.set_xlabel('Квантили нормального распределения')
    ax.set_ylabel('Значения')
    return ax


def qq_panels(df, columns):   # Q–Q графики рядом
    fig, axes = plt.subplots(1, len(columns), figsize=(4 * len(columns), 4), squeeze=False)
    for ax, column in zip(axes[0], columns):
        qq_plot(df[column], column, ax=ax)
    fig.tight_layout()
    return fig


def decision(p, alpha=0.05):   # решение словами
    if not 0 <= p <= 1:
        raise ValueError(f'p-значение должно лежать от 0 до 1, получено {p}')
    return 'отвергаем' if p < alpha else 'не отвергаем'


def ci_mean(values, conf=0.95):   # интервал среднего
    values = values.dropna()
    return stats.t.interval(conf, len(values) - 1, loc=values.mean(), scale=stats.sem(values))


def bonferroni(pvalues, alpha=0.05):   # поправка Бонферрони
    return pvalues < alpha / len(pvalues)


def t_two(a, b):   # t Уэлча и d Коэна
    a, b = a.dropna(), b.dropna()
    result = stats.ttest_ind(a, b, equal_var=False)
    d = (a.mean() - b.mean()) / np.sqrt((a.var() + b.var()) / 2)
    return result.statistic, result.pvalue, d


def mw_two(a, b):   # критерий Манна–Уитни
    result = stats.mannwhitneyu(a.dropna(), b.dropna())
    return result.statistic, result.pvalue


def chi_square(df, a, b):   # хи-квадрат и V Крамера
    table = pd.crosstab(df[a], df[b])
    result = stats.chi2_contingency(table)
    plain = stats.chi2_contingency(table, correction=False)
    v = np.sqrt(plain.statistic / (table.to_numpy().sum() * (min(table.shape) - 1)))
    return result.statistic, result.pvalue, result.dof, v


def corr_test(x, y):   # корреляции Пирсона и Спирмена
    both = x.notna() & y.notna()
    pearson = stats.pearsonr(x[both], y[both])
    spearman = stats.spearmanr(x[both], y[both])
    return pearson.statistic, pearson.pvalue, spearman.statistic, spearman.pvalue


def groups_test(df, value, group):   # ANOVA Уэлча и Крускал–Уоллис
    samples = [values.dropna() for name, values in df.groupby(group)[value]]
    anova = stats.f_oneway(*samples, equal_var=False)
    kruskal = stats.kruskal(*samples)
    return pd.Series({'F': anova.statistic, 'p ANOVA': anova.pvalue, 'H': kruskal.statistic, 'p Крускала': kruskal.pvalue})


def effect_label(value, kind):   # размер эффекта словом
    thresholds = {'d': (0.2, 0.5, 0.8), 'V': (0.1, 0.3, 0.5), 'r': (0.1, 0.3, 0.5)}
    if kind not in thresholds:
        raise ValueError(f"kind должен быть 'd', 'V' или 'r', получено {kind!r}")
    if np.isnan(value):
        raise ValueError('размер эффекта — nan: в данных пропуски')
    small, medium, large = thresholds[kind]
    size = abs(value)
    if size >= large:
        return 'большой'
    if size >= medium:
        return 'средний'
    if size >= small:
        return 'малый'
    return 'ничтожный'


def hypotheses_report(df, drivers):   # таблица гипотез курса
    experience, health, gender = drivers['experience_clean'], drivers['health'], drivers['gender']
    t1, p1, d1 = t_two(experience[health == 'Погиб'], experience[health == 'Ранен'])
    year = df['datetime'].dt.year
    by_year = pd.crosstab(df['district'], year)
    by_year = by_year[~by_year.index.str.contains('область')]
    paired = stats.ttest_rel(by_year[year.min()], by_year[year.max()])
    flags = df.assign(dark=df['light'].str.contains('темное'), fatal=df['dead_count'] > 0, defects=df['road_defects'] > 0)
    chi3, p3, dof3, v3 = chi_square(flags, 'dark', 'fatal')
    r4, p4, rho4, p_rho4 = corr_test(experience, drivers['vehicle_age'])
    t5, p5, d5 = t_two(experience[gender == 'Женский'], experience[gender == 'Мужской'])
    chi6, p6, dof6, v6 = chi_square(flags, 'defects', 'fatal')
    rows = [
        ('стаж погибших и раненых водителей', 't Уэлча', experience[health.isin(['Погиб', 'Ранен'])].count(), t1, p1, d1, effect_label(d1, 'd')),
        ('ДТП по районам в первый и последний год', 'парный t', len(by_year), paired.statistic, paired.pvalue, np.nan, '—'),
        ('тёмное время и ДТП с погибшими', 'хи-квадрат', len(df), chi3, p3, v3, effect_label(v3, 'V')),
        ('стаж и возраст машины', 'Пирсон', (experience.notna() & drivers['vehicle_age'].notna()).sum(), r4, p4, r4, effect_label(r4, 'r')),
        ('стаж женщин и мужчин за рулём', 't Уэлча', experience[gender.isin(['Женский', 'Мужской'])].count(), t5, p5, d5, effect_label(d5, 'd')),
        ('недостатки дороги и ДТП с погибшими', 'хи-квадрат', len(df), chi6, p6, v6, effect_label(v6, 'V')),
    ]
    table = pd.DataFrame(rows, columns=['вопрос', 'тест', 'n', 'статистика', 'p', 'эффект', 'размер'], index=range(1, 7))
    table['решение'] = np.where(table['p'] < 0.05, 'отвергаем', 'не отвергаем')
    table['после поправки'] = np.where(table['p'] < 0.05 / len(table), 'отвергаем', 'не отвергаем')
    return table
```

## 1. Загрузка и первичный осмотр

**Что здесь проверяет рубрика:** данные загружены верно, есть таблица с размером, памятью и повторами и таблица типов; этап 2 — 10 %, на защите — строка «Данные и качество».

## 2. Анализ пропущенных значений

**Что здесь проверяет рубрика:** полный анализ пропусков с визуализацией, скрытыми кодами и обоснованным решением; этап 2 — 15 %.

**Частая ошибка:** коды пропуска остаются числами: стаж 95, 96 и 99 — служебные коды, и с ними средний стаж участника 23,27 года вместо 15,95. Вторая — доля пропусков по всей таблице выдаётся за долю в группе, о которой вопрос: стажа нет у пешехода по смыслу, а не по недосмотру.

### 2.1. Подсчёт пропусков

```python
print(dtp.isna().sum()[dtp.isna().sum() > 0])                            # пропуски в таблице ДТП
missing = pd.DataFrame({'пропусков': people.isna().sum(), 'доля': people.isna().mean().round(3)})   # пропуски участников
missing[missing['пропусков'] > 0].sort_values('доля', ascending=False)
```

В таблице ДТП пропуски только в `road_defect` — 6 815 ДТП, две трети. В таблице участников больше всего пропусков у стажа — 9 286, 37,6 %, затем у года выпуска, марки, класса и номера машины — от 11,9 до 16,5 %, у пола — 5 %, у здоровья — 94 строки. Но это видимые пропуски; скрытые видны в `value_counts`.

```python
print(people['experience'].value_counts().sort_index().tail(5))          # самые большие значения стажа
people['experience_clean'] = people['experience'].replace([95, 96, 99], np.nan)   # коды стажа пропусками
print(people['experience'].isin([95, 96, 99]).sum(), round(people['experience'].mean(), 2), round(people['experience_clean'].mean(), 2))
```

После 69 лет значений нет до 95, а значения 95, 96 и 99 встречаются 856, 63 и 484 раза: это служебные коды, а не годы. 1 403 значения заменены пропуском в новом столбце `experience_clean`, исходный столбец не тронут. Средний стаж участника без кодов — 15,95 года вместо 23,27. Что означает каждый код, по таблице не установить, поэтому коды — пропуск, а не ноль.

### 2.2. Визуализация пропусков

**Вывод.** После замены кодов стаж неизвестен у 43 % участников, доля 0,432, — больше, чем у любого другого столбца; у столбцов машины — года выпуска, марки, класса и номера — пропусков по 12–17 %, у пола 5 %. **Оговорка.** Это доля по всем участникам: у пешехода нет ни стажа, ни машины, и такой пропуск — не дыра в данных, а отсутствие признака; доля в нужной группе — в 2.3.

### 2.3. Анализ пропусков

```python
print((dtp['road_defect'].isna() == (dtp['road_defects'] == 0)).all())   # пропуск — недостатки не установлены
main = people[people['role'].isin(['Водитель', 'Пассажир', 'Пешеход', 'Велосипедист'])]   # четыре частые роли
missing_by_group(main, 'experience_clean', 'role').round(3)              # доля пропусков стажа
```

Пропуски связаны с ролью: стажа нет у всех пассажиров, пешеходов и велосипедистов, у водителей он неизвестен в 7,9 % случаев. Пропуск в `road_defect` стоит ровно там, где счётчик `road_defects` равен нулю: «недостатки дороги не установлены».

### 2.4. Решение по пропускам

Решения по каждому столбцу с пропусками:

- стаж и возраст машины изучаются только у водителей; 7,9 % пропусков стажа не заполняются: медиана на месте пропусков сжала бы разброс и сдвинула бы сравнение групп;
- `road_defect` не заполняется: пропуск значит «не установлены», а для расчётов есть счётчик `road_defects`;
- пол и здоровье — пропуски из источника, они остаются; доли считаются по известным значениям.

Таблица для разделов 3–6 и гипотез — водители с их ДТП, как на семинарах 11–13: возраст машины — год ДТП минус год выпуска, больше 100 лет — пропуск.

```python
drivers = people[people['role'] == 'Водитель'].merge(dtp, left_on='dtp_id', right_index=True)   # водители с их ДТП
drivers['vehicle_age'] = drivers['datetime'].dt.year - drivers['vehicle_year']   # возраст машины в годах
drivers['vehicle_age'] = drivers['vehicle_age'].where(drivers['vehicle_age'] <= 100)   # невозможный возраст в пропуск
print(len(drivers), drivers['experience_clean'].count(), drivers['vehicle_age'].count())
```

15 229 водителей; стаж известен у 14 028, возраст машины — у 14 784.

## 3. Описательные статистики

**Что здесь проверяет рубрика:** статистики для числовых и категориальных столбцов с интерпретацией, период данных; этап 2 — 20 %.

**Частая ошибка:** среднее без оговорки: у скошенных столбцов — стажа, числа погибших и раненых — среднее тянут редкие большие значения, и рядом пишут медиану. Вторая — доля без `dropna`: `(people['gender'] == 'Женский').mean()` считает пропуск пола как «не женщина» и занижает долю женщин, а `value_counts(normalize=True)` пропуски не считает.

### 3.1. Числовые переменные

```python
numeric = ['participants_count', 'vehicles_count', 'dead_count', 'injured_count', 'road_defects']   # счётчики ДТП
summary = dtp[numeric].describe()
summary.loc['асимметрия'] = dtp[numeric].skew()                          # скошенность распределения
summary.round(2)
```

Типичное ДТП — два участника, одна-две машины, погибших нет, один раненый: это медианы, строка 50 %. Средние выше медиан, а асимметрия от 1,43 до 7,55: у всех счётчиков длинный правый хвост — редкие ДТП, где участников до 20, а погибших до 17. Самое тяжёлое ДТП — 17 погибших и 20 участников; это настоящее событие, разбор в разделе 6.1.

```python
summary = drivers[['experience_clean', 'vehicle_age']].describe()
summary.loc['асимметрия'] = drivers[['experience_clean', 'vehicle_age']].skew()   # скошенность распределения
summary.round(2)
```

**Вывод.** Половина водителей, попавших в ДТП, водит не больше 13 лет, четверть — не больше 6; машине половины водителей не больше 10 лет. **Оговорка.** Распределения скошены вправо, асимметрия 0,81 и 0,92: средние 15,95 и 11,51 года тянут водители с большим стажем и старые машины, поэтому типичное значение — медиана. И это водители, попавшие в ДТП, а не все водители области.

### 3.2. Категориальные переменные — самостоятельно

```python
rows = {}
for name, values in [('category', dtp['category']), ('severity', dtp['severity']), ('light', dtp['light']),
                     ('weather', dtp['weather']), ('district', dtp['district']), ('road_defect', dtp['road_defect']),
                     ('role', people['role']), ('gender', people['gender']), ('health', people['health']),
                     ('violation', people['violation'])]:
    shares = values.value_counts(normalize=True)                         # доли без пропусков
    rows[name] = {'значений': values.nunique(), 'самое частое': shares.index[0], 'его доля': round(shares.iloc[0], 3),
                  'доля трёх частых': round(shares.head(3).sum(), 3), 'редких меньше 1 %': (shares < 0.01).sum()}
pd.DataFrame.from_dict(rows, orient='index')                             # сводка категорий
```

Три самых частых вида ДТП — столкновение, наезд на пешехода, съезд с дороги — 74,9 % всех ДТП, столкновение одно — 39,5 %; в ясную погоду — 52,1 % ДТП, при трёх самых частых погодах — 96,9 %. Редкие категории есть в большинстве столбцов: 9 видов ДТП из 17 и 49 нарушений из 60 встречаются реже чем в 1 % случаев, на графиках их объединяют или оставляют за кадром. Самое частое значение района — «Брянская область», 35,7 % ДТП: это не район, а ДТП без указания района, по координатам — город Брянск, и город с районами не сравнивают. Доли посчитаны по известным значениям: у пола без пропусков мужчин 71,6 %.

## 4. Корреляционный анализ

**Что здесь проверяет рубрика:** матрица корреляций числовых столбцов, сильные пары выделены и объяснены; этап 2 — 15 %.

**Частая ошибка:** корреляцию называют причиной или называют связью r около нуля, потому что p маленькое: на тысячах пар значим и ничтожный r. Вторая — `corr()` по всей таблице: с текстовыми столбцами pandas падает, нужен список числовых столбцов по смыслу.

### 4.1. Корреляционная матрица

```python
corr_heat(dtp, numeric, 'Корреляции счётчиков ДТП');                     # тепловая карта
```

**Вывод.** Заметнее всего связаны число участников и число раненых, 0,68, и число участников и машин, 0,50; погибшие и раненые связаны слабо и отрицательно, −0,17, а число недостатков дороги не связано ни с одним счётчиком. **Оговорка.** Раненые — часть участников, а машины — места для участников: эти связи заложены в самих счётчиках и ничего не говорят о причинах ДТП.

### 4.2. Анализ корреляций

```python
corr = dtp[numeric].corr()
upper = np.triu(np.ones(corr.shape, dtype=bool), k=1)                    # каждая пара один раз
pairs = corr.where(upper).stack().dropna().sort_values()                 # пары по возрастанию
print(pairs[pairs.abs() > 0.7])                                          # сильные связи
print(round(drivers['experience_clean'].corr(drivers['vehicle_age']), 3))   # стаж и возраст машины
pairs.round(3)
```

Сильных корреляций, |r| > 0,7, нет: список пуст. Средних, от 0,4 до 0,7, две — 0,684 и 0,505, остальные слабые. Это шкала разведочного анализа: она отбирает заметные связи в таблице. Размер эффекта гипотезы в разделе Г называют по порогам Коэна — 0,1, 0,3 и 0,5, семинар 13, — и по ним 0,684 — большая связь; шкалы разные, поэтому в отчёте пишут, по какой из них связь «средняя» или «большая». Стаж водителя и возраст его машины не связаны: r −0,03. Отсутствие связи — тоже находка: опытные водители не ездят ни на новых, ни на старых машинах чаще других. Значимость этого r проверяет гипотеза 4 в разделе Г.

## Г. Проверка гипотез

### Г3. Две категории: тёмное время суток и ДТП с погибшими

```python
flags = dtp.assign(dark=np.where(dtp['light'].str.contains('темное'), 'тёмное время', 'светлое время и сумерки'))   # время суток словами
print(pd.crosstab(flags['dark'], flags['fatal'], margins=True, margins_name='всего'))   # счётчики ДТП
chi, p, dof, v = chi_square(flags, 'dark', 'fatal')                      # хи-квадрат и V Крамера
print(round(chi, 2), p, dof, round(v, 3), effect_label(v, 'V'))
mean_bars(flags.sort_values('dark'), 'dark', 'fatal', 'Доля ДТП с погибшими в тёмное и светлое время', 'Время суток', 'Доля ДТП с погибшими')
flags.groupby('dark')['fatal'].mean().round(3)                           # доли по строкам
```

- **Вопрос.** Чаще ли ДТП заканчиваются гибелью людей в тёмное время суток?
- **H0 и H1.** H0: тёмное время и гибель людей в ДТП независимы, доля ДТП с погибшими в тёмное и в светлое время одинакова; H1: признаки связаны; альтернатива двусторонняя.
- **Тест и почему.** Хи-квадрат независимости: оба признака категориальные, таблица 2 × 2 по счётчикам `pd.crosstab`; тёмное время — три вида освещения «в темное время суток», светлое — день и сумерки; в клетках от 676 до 5 946 ДТП, ожидаемые при независимости тоже намного больше пяти; p — с поправкой Йейтса, V — по χ² без неё.
- **n.** 10 114 ДТП: 3 485 в тёмное время, 6 629 в светлое время и сумерки.
- **Статистика и p.** χ² = 161,63, p < 0,001.
- **Размер эффекта.** Доля ДТП с погибшими — 19,4 % в тёмное время и 10,3 % в светлое время и сумерки, усы интервалов на графике далеко друг от друга; V Крамера 0,127 — малая связь по порогам Коэна, но разница почти вдвое важна и при малом V.
- **Вывод.** Отвергаем: в тёмное время ДТП заканчиваются гибелью людей почти вдвое чаще, чем в светлое время и сумерки. Это связь, а не причина: темнота, скорость на загородных трассах и трезвость водителей здесь не разделены, а внутри тёмного времени доля зависит от освещения — 12,1 % при включённом свете и 28,6 % без освещения, раздел 5.4.

### Г4. Две группы: стаж женщин и мужчин за рулём

```python
women = drivers.loc[drivers['gender'] == 'Женский', 'experience_clean'].dropna()   # стаж женщин за рулём
men = drivers.loc[drivers['gender'] == 'Мужской', 'experience_clean'].dropna()     # стаж мужчин за рулём
t, p, d = t_two(women, men)                                              # t Уэлча и d Коэна
weighted = np.sqrt(((len(women) - 1) * women.var() + (len(men) - 1) * men.var()) / (len(women) + len(men) - 2))   # отклонение с весами групп
print(len(women), len(men), round(t, 2), p, round(d, 3), round((women.mean() - men.mean()) / weighted, 3))
print(round(women.mean(), 2), round(men.mean(), 2), round(women.std(), 2), round(men.std(), 2))   # средние и разброс
print(women.quantile([0.25, 0.5, 0.75]).tolist(), men.quantile([0.25, 0.5, 0.75]).tolist())   # квартили и медианы
low_w, high_w = ci_mean(women)                                           # интервал среднего женщин
low_m, high_m = ci_mean(men)                                             # интервал среднего мужчин
print(round(low_w, 2), round(high_w, 2), round(low_m, 2), round(high_m, 2))
ax = box_by(drivers, 'experience_clean', 'gender', ['Женский', 'Мужской'], 'Стаж водителей по полу')   # ящики двух групп
ax.set_xlabel('Пол водителя')
ax.set_ylabel('Стаж, лет');
```

- **Вопрос.** Отличается ли стаж женщин и мужчин, попавших в ДТП за рулём?
- **H0 и H1.** H0: средний стаж женщин и мужчин одинаков, μ₁ = μ₂; H1: различается, μ₁ ≠ μ₂; альтернатива двусторонняя — направление из вопроса не следует.
- **Тест и почему.** t-тест Уэлча: вопрос о средних; нормальность стажа отвергнута, Г1, но в группах 1 865 и 11 872 значения, и по центральной предельной теореме средние распределены почти нормально; группы разного размера и разброса — отклонения 6,69 и 12,33 года, поэтому Уэлч, а не Стьюдент.
- **n.** 1 865 женщин и 11 872 мужчины с известными полом и стажем.
- **Статистика и p.** t = −45,08, p < 0,001.
- **Интервал и размер эффекта.** Средний стаж женщин — 8,44 года, 95-процентный интервал от 8,14 до 8,75; мужчин — 17,09, от 16,87 до 17,32; медианы 7 и 15 лет. d Коэна −0,872 — большой эффект, но у границы со средним: курс делит разницу средних на √((s₁² + s₂²) / 2), а классическое объединённое отклонение с весами по размерам групп даёт d −0,74 — средний эффект: у мужчин больше и группа, и разброс. Слово здесь зависит от формулы, поэтому рядом с ним пишут саму разницу — почти вдвое.
- **Вывод.** Отвергаем: женщины, попавшие в ДТП за рулём, в среднем вдвое менее опытны, чем мужчины, — 8,4 года против 17,1; три четверти женщин водят не больше 12 лет, а это меньше медианы мужчин, 15 лет. Это водители, попавшие в ДТП, а не все водители области: если среди женщин за рулём вообще больше начинающих, их будет больше и в ДТП; вывод «женщины водят хуже» из этой гипотезы не следует.

## 7. Выводы

**Что здесь проверяет рубрика:** выводы конкретны и опираются на результаты: оценка качества, ключевые находки, рекомендации; этап 2 — 10 %, на защите — строка «Выводы и ответы на вопросы», 20 %.

**Частая ошибка:** выводы пересказывают код — «построили гистограмму, посчитали корреляцию» — или вводят находку, которой нет в разделах. Вторая — рекомендация «собрать больше данных» без того, каких именно и зачем.

### 7.1. Качество данных

Качество хорошее, с оговорками. Ключевые столбцы полные: в таблице ДТП пропуск только в `road_defect`, и он значит «недостатки не установлены»; повторов ДТП нет, счётчики согласованы. Проблемы и что с ними сделано:

- стаж записан с кодами 95, 96 и 99 — 1 403 значения переведены в пропуск; у водителей стаж неизвестен в 7,9 % случаев;
- учёт лёгких ДТП менялся: 576 в 2015 году, ни одного в 2020-м, 46 в 2025-м — числа и доли разных лет сравниваются с оговоркой;
- 367 ДТП с координатами за рамкой области, 224 из них нулевые, — на карту идут только ДТП в рамке;
- 35,7 % ДТП записаны на строку «Брянская область» без района — по координатам это город Брянск, и с районами его не сравнивают;
- повторы участников — 2 017 строк неотличимых пассажиров и пешеходов — оставлены.

### 7.2. Ключевые находки

1. **Темнота и освещение связаны с гибелью.** В тёмное время гибелью кончается 19,4 % ДТП, в светлое время и сумерки — 10,3 %, V 0,127 — малая связь (гипотеза 3, Г3); без освещения — 28,6 %, при включённом освещении — 12,1 %, почти как днём, 10,1 %.
2. **Рост доли — от учёта, а не от опасности.** ДТП в таблице стало втрое меньше, доля ДТП с погибшими выросла с 0,137 до 0,215, а самих ДТП с погибшими стало меньше — 191 и 100.
3. **Погибшие водители опытнее раненых:** медианы стажа 16 и 11 лет, d 0,345 — малый эффект (гипотеза 1).
4. **Женщины за рулём в ДТП вдвое менее опытны мужчин:** 8,4 года против 17,1, d −0,872 — большой эффект, у границы со средним (гипотеза 5, Г4).
5. **Чего в данных не видно:** сильных корреляций между счётчиками ДТП нет; стаж не связан с возрастом машины, r −0,03; связь недостатков дороги с гибелью людей не обнаружена — p 0,219, V 0,013 (гипотеза 6).

### 7.3. Рекомендации

- **Перспективные столбцы:** освещение, час и вид ДТП — для вопросов о гибели людей; стаж и пол — для вопросов о водителях.
- **Предобработка:** для сравнения лет брать тяжёлые ДТП и ДТП с погибшими без лёгких; для анализа по месту — заменить координаты за рамкой пропуском.
- **Гипотезы дальше:** изменилось ли число ДТП с погибшими, а не всех ДТП, за 2015–2025 годы; отличается ли стаж водителей-нарушителей от стажа водителей без нарушений; связано ли освещение с гибелью внутри одного вида ДТП — наезда на пешехода.
- **Дополнительные данные:** интенсивность движения и протяжённость дорог по районам — чтобы перейти от числа ДТП к риску; число водителей по полу и стажу в области — чтобы сравнивать попавших в ДТП со всеми водителями; освещённость участков дорог.

## Итоги

Отчёт — это разделы этапа 2 с его нумерацией, раздел гипотез по формату этапа 3 и выводы. Под каждым графиком и тестом — вывод с числом и оговорка: что из этого не следует. Качество данных и журнал решений идут до выводов и до гипотез: читатель сначала узнаёт, можно ли данным верить. Гипотезы вырастают из находок разведочного анализа; в сводную таблицу идут все проверенные гипотезы, при больше чем пяти — с поправкой Бонферрони, а рядом с p — размер эффекта словом; нормальность — отдельная проверка вне таблицы, по ней выбирают тесты. Две главные гипотезы записываются по семи пунктам. Выводы — числа из разделов, а не пересказ кода. Функции курса — инструменты: подпись, поставленную по умолчанию, поправляют после вызова.
