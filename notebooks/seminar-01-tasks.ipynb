```python
def remove_adjacent_tests():
    assert remove_adjacent_solution([1, 1, 2, 2, 2, 3, 4, 3, 3, 5]) == [1, 2, 3, 4, 3, 5]
    assert remove_adjacent_solution([5, 5, 5]) == [5]
    assert remove_adjacent_solution([1, 2, 3]) == [1, 2, 3]
    assert remove_adjacent_solution([]) == []

    rng = np.random.default_rng(2)
    example_1 = rng.integers(1, 6, size=100).tolist()
    result_1 = remove_adjacent_solution(example_1)
    assert len(result_1) == 80
    assert result_1[:12] == [5, 2, 1, 2, 3, 5, 3, 1, 2, 4, 5, 4]
    assert result_1[-6:] == [3, 1, 2, 1, 3, 1]
    for i in range(1, len(result_1)):
        assert result_1[i] != result_1[i - 1]

    example_2 = rng.integers(1, 6, size=1000).tolist()
    result_2 = remove_adjacent_solution(example_2)
    assert len(result_2) == 781
    assert result_2[:12] == [3, 4, 3, 2, 5, 2, 5, 3, 2, 4, 5, 4]
    assert result_2[-6:] == [4, 3, 2, 5, 4, 5]
    for i in range(1, len(result_2)):
        assert result_2[i] != result_2[i - 1]

    print('Все тесты прошли успешно!')
```

## Разбиение строки по разделителю

Задача необязательная и выполняется дома.

```python
def split_line_solution(line, sep):
    """
    Разбивает строку по разделителю и возвращает список значений между
    разделителями. Если значение обособлено двойными кавычками, разделитель
    внутри него не считается разделителем, а сами кавычки в результат не входят.

    Например, для строки 'Иванов Иван,А40-1/2026,"иск, встречный иск",2026'
    и разделителя ',' результат ['Иванов Иван', 'А40-1/2026', 'иск, встречный иск', '2026'].
    Для пустой строки результат [''] — список из одной пустой строки.

    Аргументы:
        line: Строка, которую нужно разбить; двойные кавычки в ней встречаются
              только парами, обособляющими целое значение.
        sep: Разделитель, строго один символ; гарантируется, что это не двойная кавычка.

    Возвращаемое значение:
        Список значений, полученных разбиением строки по разделителю.
    """

    pass
```

```python
def split_line_tests():
    example_1_line = 'Иванов Иван,А40-1/2026,"иск, встречный иск",2026'
    example_1_res = ['Иванов Иван', 'А40-1/2026', 'иск, встречный иск', '2026']

    assert split_line_solution(example_1_line, ',') == example_1_res

    example_2_line = 'Петров Пётр;;2025'
    example_2_res = ['Петров Пётр', '', '2025']

    assert split_line_solution(example_2_line, ';') == example_2_res

    example_3_line = 'Сидорова Анна,"[запись 1, запись 2, запись 3]", ,2'
    example_3_res = ['Сидорова Анна', '[запись 1, запись 2, запись 3]', ' ', '2']

    assert split_line_solution(example_3_line, ',') == example_3_res

    example_4_line = 'А40-1/2026,,'
    example_4_res = ['А40-1/2026', '', '']

    assert split_line_solution(example_4_line, ',') == example_4_res

    example_5_line = 'А40-7/2026,"неустойка, проценты"'
    example_5_res = ['А40-7/2026', 'неустойка, проценты']

    assert split_line_solution(example_5_line, ',') == example_5_res

    example_6_line = 'А40-1/2026,"",2026'
    example_6_res = ['А40-1/2026', '', '2026']

    assert split_line_solution(example_6_line, ',') == example_6_res

    assert split_line_solution('', ',') == ['']

    print('Все тесты прошли успешно!')
```

```python
split_line_tests()
```
