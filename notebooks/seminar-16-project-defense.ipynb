## Сетка времени

Пара — 90 минут. Основная сетка — из задания на защиту, сокращённая — для большой группы; какая действует, преподаватель объявляет после подсчёта группы, до семинара 16, чтобы выступление успели прогнать с часами.

| Сетка | Выступление | Вопросы | Одна защита | Защит за пару | Когда |
|---|---|---|---|---|---|
| основная | 5 минут | 2 минуты | 7 минут | 12 — 84 минуты из 90 | до 12 выступающих |
| сокращённая | 4 минуты | 1 минута | 5 минут | 18 — все 90 минут | от 13 до 18 выступающих |

В основной сетке остаётся шесть минут на начало пары и смену выступающих. В сокращённой запаса нет: следующий выступающий начинает, как только закончились вопросы предыдущему. План выступления в ней тот же, только «что сделали бы дальше» — одной фразой в конце четвёртой минуты, а на вопросы — минута.

Ячейка ниже печатает для каждой защиты время начала выступления и вопросов. Время начала пары в вызовах — пример: подставьте своё по расписанию группы.

```python
from datetime import datetime, timedelta


def defense_schedule(speakers, talk_min, questions_min, start):
    slot = timedelta(minutes=talk_min + questions_min)                    # одна защита целиком
    first = datetime.strptime(start, '%H:%M')                             # начало пары
    print(f'Выступающих: {speakers}; выступление {talk_min} мин, вопросы {questions_min} мин')
    print('  №  начало   вопросы')
    for number in range(1, speakers + 1):
        talk = first + (number - 1) * slot
        questions = talk + timedelta(minutes=talk_min)
        print(f'{number:3}  {talk:%H:%M}    {questions:%H:%M}')
    end = first + speakers * slot                                         # конец последней защиты
    pair_end = first + timedelta(minutes=90)                              # пара — 90 минут
    print(f'Конец защит {end:%H:%M}, конец пары {pair_end:%H:%M}')


defense_schedule(12, 5, 2, '9:00')                                        # основная сетка: 5 + 2
print()
defense_schedule(18, 4, 1, '9:00')                                        # большая группа: 4 + 1
```

## Рубрика защиты

Оценивается итоговый отчёт и выступление; вес защиты в итоговой оценке объявляет преподаватель. Рубрика — раздел «Критерии оценки» задания на защиту:

| Критерий | Баллы | Описание |
|----------|-------|----------|
| **Данные и качество** | 20% | Источник, период, что в строке и сколько строк названы; вопрос исследования сформулирован; пропуски, коды, выбросы и повторы найдены, у каждой находки в журнале решение с причиной; в выступлении качество данных — одной фразой |
| **Графики** | 30% | Два главных графика отвечают на вопрос исследования; у них заголовки и подписи осей; под каждым графиком отчёта — вывод с числом и оговорка; в отчёте не меньше пяти разных графиков |
| **Гипотезы** | 30% | Две гипотезы выступления записаны по семи пунктам формата этапа 3; тест выбран по нормальности и размеру выборки; рядом с p — размер эффекта словом; решение — «отвергаем» или «не отвергаем»; в отчёте все гипотезы этапа 3, при больше чем пяти — с поправкой Бонферрони |
| **Выводы и ответы на вопросы** | 20% | Выводы раздела 7 опираются на числа отчёта и не сильнее данных; «что сделали бы дальше» — конкретно: какие данные и какой вопрос; ответы на вопросы по существу; свой вопрос по чужому отчёту задан |

**Штрафные баллы:**
- **-10%**: Ноутбук не выполняется сверху вниз или содержит критические ошибки
- **-5%**: Вывод сильнее данных: «доказали», «приняли», связь названа причиной
