## Как проходит контрольная

- Задания выдаёт тест Moodle «КР 2»: 80 минут, одна попытка, пароль объявляется в начале пары.
- В тесте девять заданий. Версии заданий с разными числами и параметрами выдаются случайно, поэтому у соседа, как правило, другие задания и другие ответы.
- Считаете в этом ноутбуке, ответы вводите в тест: числа, названия и решения тестов проверяются автоматически. Функцию из задания 9 пишете в ноутбуке, в тест вводите числа по двум её графикам; код в тест не вставляется.
- Дробные числа вводите с точкой, как их печатает Python: 3.14, а не 3,14; в коде разделитель тоже точка. Округляйте так, как сказано в условии; решение теста выбирается из списка «отвергаем» или «не отвергаем».
- Значения из таблицы, которые нужны в коде, — названия районов, категорий и марок, примеры вызовов функции — даны под условием блоком кода. Копируйте из блока, а не из текста условия: вместе с текстом могут скопироваться невидимые символы форматирования. Названия столбцов набирайте с клавиатуры.
- Ответ теста должен получаться из кода ноутбука. Работа, в которой ответы есть, а кода нет, не засчитывается.
- Графики в тесте не проверяются: по графику спрашивается число, которое считается по той же сводке. Постройте график в ноутбуке — он остаётся в сданном файле.
